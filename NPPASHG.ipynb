```python
# anp_radiation.py
import math
import time
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
import matplotlib.pyplot as plt
import pandas as pd
from tqdm import tqdm
import os

from google.colab import drive
drive.mount('/content/drive')

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
R_EARTH = 6371000.0  # meters

def save_checkpoint(model, optimizer, epoch, path):
    """
    Save model + optimizer + epoch
    """
    os.makedirs(os.path.dirname(path), exist_ok=True)
    torch.save({
        'epoch': epoch,
        'model_state_dict': model.state_dict(),
        'optimizer_state_dict': optimizer.state_dict(),
    }, path)
    print(f"Saved checkpoint: {path}")

# ---------------------------
# Utils
# ---------------------------

def latlon_to_local_meters(lat, lon, latc, lonc):
    lat = np.asarray(lat, dtype=float)
    lon = np.asarray(lon, dtype=float)
    latc = float(latc)
    lonc = float(lonc)
    lat_rad = np.deg2rad(lat)
    lon_rad = np.deg2rad(lon)
    latc_rad = math.radians(latc)
    lonc_rad = math.radians(lonc)
    dx = (lon_rad - lonc_rad) * np.cos(latc_rad) * R_EARTH
    dy = (lat_rad - latc_rad) * R_EARTH
    return dx, dy

# ---------------------------
# Dataset
# ---------------------------

class RadiationDataset(Dataset):
    def __init__(self, data_np, example_size=200, log_transform=True):
        self.data = data_np.copy()
        self.example_size = example_size
        if log_transform:
            self.data[:,3] = np.log1p(self.data[:,3])
        self.N = len(self.data)

    def __len__(self):
        return max(1000, self.N // 10)

    def __getitem__(self, idx):
        inds = np.random.choice(self.N, size=self.example_size, replace=False)
        lat, lon, t, v = self.data[inds].T
        v = np.log1p(v)
        latc, lonc, tc = lat.mean(), lon.mean(), t.mean()
        dx, dy = latlon_to_local_meters(lat, lon, latc, lonc)
        dt = t - tc

        dx = dx / 50000.0
        dy = dy / 50000.0
        dt = dt / 86400.0

        x = np.stack([dx, dy, dt], axis=-1).astype(np.float32)
        y = v.astype(np.float32)

        assert np.isfinite(x).all(), f"x contains non-finite values at index {idx}"
        assert np.isfinite(y).all(), f"y contains non-finite values at index {idx}"

        meta = {'latc': latc, 'lonc': lonc, 'tc': tc}
        return x, y, meta

# ---------------------------
# Model
# ---------------------------

class MLP(nn.Module):
    def __init__(self, in_dim, hidden, out_dim, n_layers=3):
        super().__init__()
        layers = []
        d = in_dim
        for _ in range(n_layers-1):
            layers.append(nn.Linear(d, hidden))
            layers.append(nn.ReLU())
            d = hidden
        layers.append(nn.Linear(d, out_dim))
        self.net = nn.Sequential(*layers)

        for m in self.modules():
            if isinstance(m, nn.Linear):
                nn.init.xavier_uniform_(m.weight)
                nn.init.zeros_(m.bias)

    def forward(self, x): return self.net(x)

class ANP(nn.Module):
    def __init__(self, x_dim=3, y_dim=1, r_dim=128, hidden=128, n_heads=8):
        super().__init__()
        self.encoder = MLP(x_dim + y_dim, hidden, r_dim)
        self.x_embed = MLP(x_dim, hidden, r_dim)
        self.mha = nn.MultiheadAttention(embed_dim=r_dim, num_heads=n_heads, batch_first=True)
        self.decoder = MLP(2*r_dim, hidden, 2)

    def forward(self, x_ctx, y_ctx, x_tgt):
        B, Nc, _ = x_ctx.shape
        Nt = x_tgt.shape[1]

        r_ctx = self.encoder(torch.cat([x_ctx, y_ctx.unsqueeze(-1)], dim=-1).view(B*Nc, -1)).view(B, Nc, -1)
        r_tgt = self.x_embed(x_tgt.view(B*Nt, -1)).view(B, Nt, -1)

        attn_out, _ = self.mha(r_tgt, r_ctx, r_ctx)
        dec_in = torch.cat([r_tgt, attn_out], dim=-1)
        dec_out = self.decoder(dec_in.view(B*Nt, -1)).view(B, Nt, 2)

        mu = dec_out[...,0]
        raw_log_sigma = dec_out[...,1]
        log_sigma = torch.clamp(raw_log_sigma, min=-10.0, max=5.0)
        sigma = 1e-3 + torch.exp(log_sigma)
        return mu, sigma

# ---------------------------
# Loss
# ---------------------------

def gaussian_nll(y, mu, sigma):
    var = sigma**2
    nll = 0.5*((y - mu)**2/var + 2*torch.log(sigma) + math.log(2*math.pi))
    return nll.mean()

# ---------------------------
# Helpers
# ---------------------------

def collate_fn(batch):
    xs = torch.from_numpy(np.stack([b[0] for b in batch], axis=0)).to(DEVICE)
    ys = torch.from_numpy(np.stack([b[1] for b in batch], axis=0)).to(DEVICE)
    metas = [b[2] for b in batch]
    return xs, ys, metas

def sample_context_target(xs, ys, n_ctx, n_tgt):
    B, S, xdim = xs.shape
    ctx_inds = torch.randint(0, S, (B, n_ctx), device=DEVICE)
    tgt_inds = torch.randint(0, S, (B, n_tgt), device=DEVICE)
    x_ctx = xs.gather(1, ctx_inds.unsqueeze(-1).expand(-1, -1, xdim))
    y_ctx = ys.gather(1, ctx_inds)
    x_tgt = xs.gather(1, tgt_inds.unsqueeze(-1).expand(-1, -1, xdim))
    y_tgt = ys.gather(1, tgt_inds)
    return x_ctx, y_ctx, x_tgt, y_tgt

# ---------------------------
# Training
# ---------------------------

def train_anp(dataset, model, batch_size=1024, epochs=20, lr=1e-5, device="cuda"):
    loader = DataLoader(dataset, batch_size=batch_size, shuffle=True, collate_fn=collate_fn)
    model.to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    loss_fn = nn.MSELoss()
    scaler = torch.amp.GradScaler('cuda')  # AMP용 스케일러

    checkpoint = torch.load("/content/drive/MyDrive/anp_checkpoints/anp_epoch_4.pt", weights_only=True)
    model.load_state_dict(checkpoint['model_state_dict'])
    optimizer.load_state_dict(checkpoint['optimizer_state_dict'])

    for ep in range(epochs):
        pbar = tqdm(loader)
        for batch_x, batch_y, batch_meta in pbar:
            batch_x = batch_x.to(device, non_blocking=True)
            batch_y = batch_y.to(device, non_blocking=True)

            if torch.isnan(batch_x).any() or torch.isnan(batch_y).any():
              print("Found NaN in batch, skipping this batch")
              continue

            # context / target split
            B, S, xdim = batch_x.shape

            # choose 50% as context, 50% as target
            n_ctx = S // 2
            n_tgt = S - n_ctx

            # random indices for each example in the batch
            idx = torch.randperm(S, device=batch_x.device)
            ctx_idx = idx[:n_ctx]
            tgt_idx = idx[n_ctx:]

            xc = batch_x[:, ctx_idx, :]
            yc = batch_y[:, ctx_idx]

            xt = batch_x[:, tgt_idx, :]
            yt = batch_y[:, tgt_idx]

            optimizer.zero_grad()

            # AMP autocast 적용
            with torch.amp.autocast('cuda'):
                mu, sigma = model(xc, yc, xt)  # ANP 출력
                if torch.isnan(mu).any() or torch.isnan(sigma).any():
                  print("NaN detected in model output, skipping step")
                  continue
                # 예제: mu로만 MSE 계산, 필요하면 sigma 포함 NLL 계산 가능
                loss = gaussian_nll(yt, mu, sigma)

            # backward는 scaler 사용
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()

            pbar.set_postfix({"loss": loss.item()})
        checkpoint_path = f"/content/drive/MyDrive/anp_checkpoints/anp_epoch_{ep+1}.pt"
        save_checkpoint(model, optimizer, ep+1, checkpoint_path)

    return model

# ---------------------------
# Heatmap
# ---------------------------

def generate_heatmap(model, context_x, context_y, meta, grid_m=100, extent_m=5000, delta_t=0.0):
    model.eval()
    device = next(model.parameters()).device

    xs = np.linspace(-extent_m, extent_m, grid_m)
    ys = np.linspace(-extent_m, extent_m, grid_m)
    XX, YY = np.meshgrid(xs, ys)
    pts = np.stack([XX.ravel(), YY.ravel(), np.full(XX.size, delta_t)], axis=-1).astype(np.float32)

    x_ctx = torch.tensor(context_x[None,:,:], dtype=torch.float32, device=device)
    y_ctx = torch.tensor(context_y[None,:], dtype=torch.float32, device=device)
    x_tgt = torch.tensor(pts[None,:,:], dtype=torch.float32, device=device)

    with torch.no_grad():
        mu, sigma = model(x_ctx, y_ctx, x_tgt)
        mu = mu.cpu().numpy().reshape(grid_m, grid_m)
        sigma = sigma.cpu().numpy().reshape(grid_m, grid_m)

    return XX, YY, mu, sigma

def plot_heatmap(XX, YY, mu, sigma, context_x=None, context_y=None):
    fig, ax = plt.subplots(1,2, figsize=(12,5))
    im0 = ax[0].imshow(mu, origin='lower', extent=[XX.min(), XX.max(), YY.min(), YY.max()])
    ax[0].set_title("Predicted mean")
    plt.colorbar(im0, ax=ax[0])
    im1 = ax[1].imshow(sigma, origin='lower', extent=[XX.min(), XX.max(), YY.min(), YY.max()])
    ax[1].set_title("Predicted sigma")
    plt.colorbar(im1, ax=ax[1])

    if context_x is not None and context_y is not None:
        dx = context_x[:,0]
        dy = context_x[:,1]
        ax[0].scatter(dx, dy, c='white', s=10, marker='x')
    plt.tight_layout()
    plt.show()

# ---------------------------
# Main
# ---------------------------

def main():
    # 예시 데이터: np.random.randn로 대체, 실제 데이터로 교체 가능
    df = pd.read_csv("./out/measurements.csv", on_bad_lines='skip', sep=";")

    dt_series = pd.to_datetime(df['dateAndTimeOfCreation'], utc=True, errors="coerce")

    timestamps = (dt_series.astype('int64') // 10**9).astype('float32')

    lat  = pd.to_numeric(df['latitude'],  errors='coerce').astype('float32')
    lon  = pd.to_numeric(df['longitude'],  errors='coerce').astype('float32')
    hits = pd.to_numeric(df['hitsNumber'], errors='coerce').astype('float32')

    mask = (~np.isnan(lat)) & (~np.isnan(lon)) & (~np.isnan(hits)) & (~np.isnan(timestamps))
    lat, lon, hits, timestamps = lat[mask], lon[mask], hits[mask], timestamps[mask]


    data = np.stack([lat, lon, timestamps, hits], axis=1).astype('float32')
    print("Loaded:", data.shape)

    dataset = RadiationDataset(data, example_size=200, log_transform=True)
    model = ANP(x_dim=3, y_dim=1, r_dim=128, hidden=128, n_heads=8).to(DEVICE)

    print("Start training...")
    # train_anp(dataset, model, epochs=10, batch_size=1024, lr=3e-5)
    print("Training done.")

    checkpoint = torch.load("/content/drive/MyDrive/anp_checkpoints/anp_epoch_4.pt", weights_only=True)
    model.load_state_dict(checkpoint['model_state_dict'])

    # visualize one example
    for i in range(100):
      x_pool, y_pool, meta = dataset[i]
      print(meta)
      XX, YY, mu, sigma = generate_heatmap(model, x_pool, y_pool, meta, grid_m=100, extent_m=5000)
      plot_heatmap(XX, YY, mu, sigma, context_x=x_pool, context_y=y_pool)

if __name__=="__main__":
    main()
```
